# Analisis de Red de Tiendas - RetailNow

## Descripcion
Este proyecto analiza los datos de ventas, inventarios y satisfaccion del cliente para la cadena de tiendas minoristas RetailNow. 
Utilizamos Pandas para manipulacion de datos y Numpy para calculos estadisticos y simulaciones.

## Objetivo
Optimizar el rendimiento de las tiendas mediante el analisis detallado de:
- Ventas totales por tienda
- Rotacion de inventarios
- Niveles de satisfaccion del cliente
- Proyecciones de ventas futuras

## 1. Preparar el Entorno - Importar Librerias Necesarias

```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

print('Librerias importadas exitosamente')
print('Pandas version:', pd.__version__)
print('Numpy version:', np.__version__)
```

## 2. Cargar los Datos - Lectura desde Archivos CSV

```python
# Rutas de los archivos CSV - relativas desde la carpeta del notebook
ruta_ventas = '/workspace/sales.csv'
ruta_inventarios = '/workspace/inventories.csv'
ruta_satisfaccion = '/workspace/satisfaction.csv'

# Cargar los datos en DataFrames
df_ventas = pd.read_csv(ruta_ventas)
df_inventarios = pd.read_csv(ruta_inventarios)
df_satisfaccion = pd.read_csv(ruta_satisfaccion)

print('Datos cargados exitosamente')
print('-' * 50)
```

## 3. Limpiar los Datos - Eliminar Valores Nulos

```python
# Eliminar filas con valores nulos
df_ventas = df_ventas.dropna()
df_inventarios = df_inventarios.dropna()
df_satisfaccion = df_satisfaccion.dropna()

print('\nDatos limpiados exitosamente')
print('Filas en df_ventas:', len(df_ventas))
print('Filas en df_inventarios:', len(df_inventarios))
print('Filas en df_satisfaccion:', len(df_satisfaccion))
```

```python
# === CONFIGURACION DE PARAMETROS PARA ANALISIS ===
print('\nPARAMETROS DE CONFIGURACION DEL ANALISIS:')
print('=' * 60)

# Parametros de simulacion
NUM_MESES_PROYECCION = 12   # Horizonte de proyeccion en meses
NUM_SIMULACIONES = 1000      # Numero de escenarios a simular

print(f'  Proyeccion: {NUM_MESES_PROYECCION} meses')
print(f'  Simulaciones: {NUM_SIMULACIONES} escenarios')
print('=' * 60)
```

## 4. Exploracion de Datos - Estructura y Resumen Estadistico

```python
# Mostrar las primeras filas de cada DataFrame
print('PRIMERAS FILAS DE DATOS DE VENTAS:')
print('-' * 50)
print(df_ventas.head())
print('\nESTRUCTURA DEL DATAFRAME DE VENTAS:')
print(df_ventas.info())
print('\nESTADISTICAS DESCRIPTIVAS DE VENTAS:')
print(df_ventas.describe())
```

```python
# Informacion de inventarios
print('PRIMERAS FILAS DE DATOS DE INVENTARIOS:')
print('-' * 50)
print(df_inventarios.head())
print('\nESTRUCTURA DEL DATAFRAME DE INVENTARIOS:')
print(df_inventarios.info())
print('\nESTADISTICAS DESCRIPTIVAS DE INVENTARIOS:')
print(df_inventarios.describe())
```

```python
# Informacion de satisfaccion
print('PRIMERAS FILAS DE DATOS DE SATISFACCION:')
print('-' * 50)
print(df_satisfaccion.head())
print('\nESTRUCTURA DEL DATAFRAME DE SATISFACCION:')
print(df_satisfaccion.info())
print('\nESTADISTICAS DESCRIPTIVAS DE SATISFACCION:')
print(df_satisfaccion.describe())
```

```python
# Validación de tipos de datos
print('\nVALIDACION DE TIPOS DE DATOS:')
print('=' * 60)
print('\nTipos en df_ventas:')
print(df_ventas.dtypes)
print('\nTipos en df_inventarios:')
print(df_inventarios.dtypes)
print('\nTipos en df_satisfaccion:')
print(df_satisfaccion.dtypes)

# Validación de tipos críticos
print('\nVERIFICACION DE TIPOS CRITICOS:')
print('-' * 60)

# ID_Tienda debe ser entero
if df_ventas['ID_Tienda'].dtype == 'int64' or df_ventas['ID_Tienda'].dtype == 'int32':
    print('✓ ID_Tienda en ventas: tipo CORRECTO (entero)')
else:
    print('⚠ ID_Tienda en ventas: tipo INCORRECTO (esperado int, obtuvo', df_ventas['ID_Tienda'].dtype, ')')

# Cantidad_Vendida debe ser numérico
if pd.api.types.is_numeric_dtype(df_ventas['Cantidad_Vendida']):
    print('✓ Cantidad_Vendida: tipo CORRECTO (numérico)')
else:
    print('⚠ Cantidad_Vendida: tipo INCORRECTO')

# Stock_Disponible debe ser numérico
if pd.api.types.is_numeric_dtype(df_inventarios['Stock_Disponible']):
    print('✓ Stock_Disponible: tipo CORRECTO (numérico)')
else:
    print('⚠ Stock_Disponible: tipo INCORRECTO')

# Verificar que IDs sean consistentes en todos los DataFrames
tiendas_ventas = set(df_ventas['ID_Tienda'].unique())
tiendas_inventarios = set(df_inventarios['ID_Tienda'].unique())
tiendas_satisfaccion = set(df_satisfaccion['ID_Tienda'].unique())

print('\nCONSISTENCIA DE TIENDAS ENTRE DATASETS:')
print(f'  Tiendas en ventas: {sorted(tiendas_ventas)}')
print(f'  Tiendas en inventarios: {sorted(tiendas_inventarios)}')
print(f'  Tiendas en satisfaccion: {sorted(tiendas_satisfaccion)}')

tiendas_comunes = tiendas_ventas & tiendas_inventarios & tiendas_satisfaccion
if len(tiendas_comunes) == len(tiendas_satisfaccion):
    print(f'✓ CONSISTENCIA OK: Todas las tiendas estan presentes en todos los datasets')
else:
    print(f'⚠ ALERTA: Hay inconsistencias en los datasets')
```

## 5. Analisis de Ventas - Calculos Principales

```python
# Sumar la cantidad vendida agrupada por ID_Tienda y Producto
suma_cantidad = df_ventas.groupby(['ID_Tienda', 'Producto'])['Cantidad_Vendida'].sum()
print('Suma de cantidad vendida por tienda y producto:')
print(suma_cantidad)
```

```python
# Calcular ingresos totales por tienda: sum(Cantidad_Vendida * Precio_Unitario) agrupado por ID_Tienda
df_ventas['Ingresos'] = df_ventas['Cantidad_Vendida'] * df_ventas['Precio_Unitario']
ventas_por_tienda = df_ventas.groupby('ID_Tienda')['Ingresos'].sum()
print('Ingresos totales por tienda:')
ventas_por_tienda
```

```python
# Crear DataFrame combinado: Ventas + Inventarios
print('DATAFRAME COMBINADO - INTEGRACION DE VENTAS E INVENTARIOS:')
print('-' * 50)

# Merge de ventas con inventarios usando las claves ID_Tienda y Producto
df_ventas_inventarios = pd.merge(
    df_ventas[['ID_Tienda', 'Producto', 'Cantidad_Vendida', 'Precio_Unitario', 'Ingresos']],
    df_inventarios[['ID_Tienda', 'Producto', 'Stock_Disponible']],
    on=['ID_Tienda', 'Producto'],
    how='left'
)

print('Primeros registros del DataFrame combinado:')
print(df_ventas_inventarios.head(10))

print(f'\nDimensiones del DataFrame combinado: {df_ventas_inventarios.shape}')
print(f'  Filas: {len(df_ventas_inventarios)} registros')
print(f'  Columnas: {len(df_ventas_inventarios.columns)} campos')

print('\nColumnas disponibles en el DataFrame combinado:')
print(f'  {", ".join(df_ventas_inventarios.columns.tolist())}')

print('\nEstructura del DataFrame combinado:')
print(df_ventas_inventarios.info())
```

```python
df_ventas.describe()
```

```python
# Resumen estadistico detallado de ingresos por tienda
print('RESUMEN ESTADISTICO DE INGRESOS POR TIENDA:')
print('-' * 50)

resumen_ventas = df_ventas.groupby('ID_Tienda')['Ingresos'].describe()
print(resumen_ventas)

print('\nANALISIS POR PRODUCTO:')
print('-' * 50)
ventas_por_producto = df_ventas.groupby('Producto')['Ingresos'].agg(['sum', 'count', 'mean']).sort_values('sum', ascending=False)
ventas_por_producto.columns = ['Ingresos_Total', 'Cantidad_Transacciones', 'Ingreso_Promedio']
print(ventas_por_producto)
```

## 6. Analisis de Inventarios - Rotacion y Niveles Criticos

```python
# Calcular rotacion de inventarios - Merge de ventas e inventarios
print('CALCULO DE ROTACION DE INVENTARIOS:')
print('-' * 50)

# Agrupar ventas por tienda y producto
ventas_agrupadas = df_ventas.groupby(['ID_Tienda', 'Producto'])['Cantidad_Vendida'].sum().reset_index()
ventas_agrupadas.columns = ['ID_Tienda', 'Producto', 'Total_Cantidad_Vendida']

# Hacer merge con inventarios usando ID_Tienda y Producto
df_rotacion = pd.merge(ventas_agrupadas, df_inventarios, on=['ID_Tienda', 'Producto'], how='left')

# Calcular rotacion de inventarios: Total_Cantidad_Vendida / Stock_Disponible
df_rotacion['Rotacion_Inventario'] = df_rotacion['Total_Cantidad_Vendida'] / df_rotacion['Stock_Disponible']

# Convertir a porcentaje
df_rotacion['Rotacion_Porcentaje'] = df_rotacion['Rotacion_Inventario'] * 100

print('Primeros registros de rotacion de inventarios:')
print(df_rotacion[['ID_Tienda', 'Producto', 'Total_Cantidad_Vendida', 'Stock_Disponible', 'Rotacion_Porcentaje']].head(15))
```

```python
# Resumen de rotacion por tienda
print('PROMEDIO DE ROTACION DE INVENTARIOS POR TIENDA:')
print('-' * 50)

rotacion_por_tienda = df_rotacion.groupby('ID_Tienda')['Rotacion_Porcentaje'].agg(['mean', 'min', 'max', 'count']).sort_values('mean', ascending=False)
rotacion_por_tienda.columns = ['Rotacion_Promedio_%', 'Rotacion_Minima_%', 'Rotacion_Maxima_%', 'Cantidad_SKU']
print(rotacion_por_tienda)

print('\nREPORTE DE TIENDAS CON PEOR ROTACION:')
print('-' * 50)
tiendas_bajo_rendimiento = rotacion_por_tienda.sort_values('Rotacion_Promedio_%').head(5)
print('Top 5 tiendas con menor rotacion de inventarios:')
for tienda, row in tiendas_bajo_rendimiento.iterrows():
    print(f'Tienda {tienda}: Rotacion promedio {row["Rotacion_Promedio_%"]:.2f}% - {int(row["Cantidad_SKU"])} productos - Rango: {row["Rotacion_Minima_%"]:.2f}% a {row["Rotacion_Maxima_%"]:.2f}%')
```

```python
# Filtrar tiendas con inventarios criticos (menos del 10% de rotacion)
print('TIENDAS CON INVENTARIOS CRITICOS (< 10% de ROTACION):')
print('-' * 50)

df_critico = df_rotacion[df_rotacion['Rotacion_Porcentaje'] < 10].copy()

if len(df_critico) > 0:
    print(f'Se encontraron {len(df_critico)} registros con inventario critico')
    print('\nDetalle de items criticos:')
    df_critico_ordenado = df_critico.sort_values(['ID_Tienda', 'Rotacion_Porcentaje'])
    print(df_critico_ordenado[['ID_Tienda', 'Producto', 'Total_Cantidad_Vendida', 'Stock_Disponible', 'Rotacion_Porcentaje']].to_string())
    
    print('\n\nRESUMEN DE TIENDAS CON ARTICULOS CRITICOS:')
    tiendas_criticas = df_critico.groupby('ID_Tienda').agg({
        'Producto': 'count',
        'Rotacion_Porcentaje': ['mean', 'min']
    }).round(2)
    tiendas_criticas.columns = ['Cantidad_Articulos_Criticos', 'Rotacion_Promedio_%', 'Rotacion_Minima_%']
    print(tiendas_criticas.sort_values('Rotacion_Minima_%'))
    
    print('\n\nRECOMENDACIONES:')
    print('- Revisar politica de abastecimiento para estos productos')
    print('- Considerar reducir stock inicial para articulos con baja rotacion')
    print('- Evaluar si estos productos son rentables o deben descontinuarse')
else:
    print('No hay tiendas con inventarios criticos. Todas tienen rotacion >= 10%')
```

## 7. Analisis de Satisfaccion del Cliente

```python
# Analizar satisfaccion del cliente por tienda
print('ANALISIS DE SATISFACCION DEL CLIENTE:')
print('-' * 50)

print('Satisfaccion por tienda:')
print(df_satisfaccion.sort_values('ID_Tienda')[['ID_Tienda', 'Satisfacción_Promedio']])

print('\nESTADISTICAS DE SATISFACCION:')
print(f'Satisfaccion promedio general: {df_satisfaccion["Satisfacción_Promedio"].mean():.2f}%')
print(f'Satisfaccion maxima: {df_satisfaccion["Satisfacción_Promedio"].max():.2f}%')
print(f'Satisfaccion minima: {df_satisfaccion["Satisfacción_Promedio"].min():.2f}%')
```

```python
# Filtrar tiendas con baja satisfaccion (< 60%)
print('TIENDAS CON BAJA SATISFACCION (< 60%):')
print('-' * 50)

satisfaccion_baja = df_satisfaccion[df_satisfaccion['Satisfacción_Promedio'] < 60].copy().sort_values('Satisfacción_Promedio')

if len(satisfaccion_baja) > 0:
    print(f'Se encontraron {len(satisfaccion_baja)} tienda(s) con baja satisfaccion:\n')
    
    for idx, row in satisfaccion_baja.iterrows():
        tienda_id = row['ID_Tienda']
        satisfaccion = row['Satisfacción_Promedio']
        
        # Calcular metricas de desempeno para esta tienda
        ventas_tienda = df_ventas[df_ventas['ID_Tienda'] == tienda_id]['Ingresos'].sum()
        transacciones_tienda = len(df_ventas[df_ventas['ID_Tienda'] == tienda_id])
        rotacion_tienda = rotacion_por_tienda.loc[tienda_id, 'Rotacion_Promedio_%'] if tienda_id in rotacion_por_tienda.index else 0
        articulos_criticos = len(df_critico[df_critico['ID_Tienda'] == tienda_id])
        
        print(f'TIENDA {tienda_id}:')
        print(f'  Satisfaccion actual: {satisfaccion:.2f}%')
        print(f'  Ingresos totales: {ventas_tienda:,.2f}')
        print(f'  Numero de transacciones: {transacciones_tienda}')
        print(f'  Rotacion promedio inventarios: {rotacion_tienda:.2f}%')
        print(f'  Articulos con inventario critico: {articulos_criticos}')
        
        # Interpretacion y recomendaciones
        print(f'\\n  INTERPRETACION:')
        if satisfaccion < 50:
            print(f'  Estado: CRITICO - Requiere intervencion inmediata')
        elif satisfaccion < 60:
            print(f'  Estado: BAJO - Requiere mejora urgente')
        
        print(f'\\n  POSIBLES CAUSAS Y RECOMENDACIONES:')
        if articulos_criticos > 0:
            print(f'  - Alto numero de articulos con inventario critico ({articulos_criticos})')
            print(f'    Recomendacion: Optimizar disponibilidad de productos')
        
        if rotacion_tienda < 20:
            print(f'  - Baja rotacion de inventarios ({rotacion_tienda:.2f}%)')
            print(f'    Recomendacion: Revisar pertinencia del surtido actual')
        
        if transacciones_tienda < 50:
            print(f'  - Bajo volumen de transacciones ({transacciones_tienda})')
            print(f'    Recomendacion: Aumentar actividades de promocion y marketing')
        
        print(f'  - Posibles problemas: Servicio al cliente, disponibilidad de productos,')
        print(f'    ambiente de tienda, atencion de empleados')
        print()
else:
    print('Excelente: No hay tiendas con baja satisfaccion. Todas estan por encima del 60%')
```

## 9. Proyecciones de Ventas Futuras - Simulacion con Numpy

```python
# CALCULOS ESTADISTICOS CON NUMPY - Directamente sobre la columna 'Ingresos'
print('CALCULOS ESTADISTICOS CON NUMPY:')
print('-' * 50)

# Convertir la columna de ingresos a array de Numpy
ingresos_array = df_ventas['Ingresos'].to_numpy()

# Calcular estadísticos con Numpy
mediana_np = np.median(ingresos_array)
desv_estandar_np = np.std(ingresos_array)
media_np = np.mean(ingresos_array)

print(f'Mediana de Ingresos (Numpy): {mediana_np:,.2f}')
print(f'Desviacion Estandar (Numpy): {desv_estandar_np:,.2f}')
print(f'Media de Ingresos (Numpy): {media_np:,.2f}')

# Calcular percentiles con Numpy
percentil_25 = np.percentile(ingresos_array, 25)
percentil_75 = np.percentile(ingresos_array, 75)
print(f'Percentil 25 (Numpy): {percentil_25:,.2f}')
print(f'Percentil 75 (Numpy): {percentil_75:,.2f}')
print(f'Rango Intercuartilico (IQR): {percentil_75 - percentil_25:,.2f}')

# Calcular coeficiente de variacion
cv = (desv_estandar_np / media_np) * 100
print(f'Coeficiente de Variacion: {cv:.2f}%')

# VALIDACIÓN: Comparar con Pandas para asegurar consistencia
print('\n' + '=' * 60)
print('VALIDACION CON PANDAS - Comparacion de resultados:')
print('=' * 60)
mediana_pd = df_ventas['Ingresos'].median()
desv_estandar_pd = df_ventas['Ingresos'].std()
media_pd = df_ventas['Ingresos'].mean()

print(f'Mediana de Ingresos (Pandas): {mediana_pd:,.2f}')
print(f'Desviacion Estandar (Pandas): {desv_estandar_pd:,.2f}')
print(f'Media de Ingresos (Pandas): {media_pd:,.2f}')

print('\nDIFERENCIAS (Numpy vs Pandas):')
diff_mediana = abs(mediana_np - mediana_pd)
diff_desv = abs(desv_estandar_np - desv_estandar_pd)
diff_media = abs(media_np - media_pd)

print(f'  Mediana: {diff_mediana:.8f}')
print(f'  Desv.Est: {diff_desv:.8f}')
print(f'  Media: {diff_media:.8f}')

if diff_mediana < 0.01 and diff_desv < 0.01 and diff_media < 0.01:
    print('\n✓ Valores CONSISTENTES entre Numpy y Pandas')
    print('✓ Los calculos estadisticos son CONFIABLES y VALIDADOS')
else:
    print('\n⚠ ALERTA: Hay diferencias significativas entre Numpy y Pandas')
```

```python
# Simular proyecciones de ventas futuras
print('SIMULACION DE PROYECCIONES DE VENTAS FUTURAS:')
print('-' * 50)

# Establecer la semilla para reproducibilidad
np.random.seed(42)

# Parametros para la simulacion
ingresos_actuales = df_ventas['Ingresos'].to_numpy()
media_ingresos = np.mean(ingresos_actuales)
desv_estandar = np.std(ingresos_actuales)
num_meses = 12  # Proyeccion para 12 meses
num_simulaciones = 1000  # Numero de simulaciones

# Generar proyecciones aleatorias usando distribucion normal
proyecciones = np.random.normal(media_ingresos, desv_estandar, (num_simulaciones, num_meses))

# Asegurar que las proyecciones sean positivas (no hay ingresos negativos)
proyecciones = np.maximum(proyecciones, 0)

print(f'Parametros de la simulacion:')
print(f'  Media de ingresos actuales: {media_ingresos:,.2f}')
print(f'  Desviacion estandar: {desv_estandar:,.2f}')
print(f'  Numero de meses proyectados: {num_meses}')
print(f'  Numero de simulaciones: {num_simulaciones}')
print(f'\\nDimensiones del array de proyecciones: {proyecciones.shape}')
print(f'\\nEsto representa {num_simulaciones} escenarios posibles de ventas durante {num_meses} meses')
```

## 10. Resumen y Recomendaciones

```python
# Analizar resultados de la simulacion
print('ANALISIS DE LAS PROYECCIONES SIMULADAS:')
print('-' * 50)

# Calcular estadisticas de las proyecciones por mes
media_proyectada = np.mean(proyecciones, axis=0)
desv_proyectada = np.std(proyecciones, axis=0)
minimo_proyectado = np.min(proyecciones, axis=0)
maximo_proyectado = np.max(proyecciones, axis=0)

# Calcular percentiles 5 y 95 para intervalos de confianza 90%
percentil_5 = np.percentile(proyecciones, 5, axis=0)
percentil_95 = np.percentile(proyecciones, 95, axis=0)

print('\\nEstadisticas de Proyecciones por Mes (Ingresos diarios promedio):')
print('Mes | Media     | Desv.Est | Minimo    | Maximo    | IC 90% (5%-95%)')
print('-' * 70)
for mes in range(num_meses):
    print(f'{mes+1:3d} | {media_proyectada[mes]:9,.0f} | {desv_proyectada[mes]:8,.0f} | {minimo_proyectado[mes]:9,.0f} | {maximo_proyectado[mes]:9,.0f} | {percentil_5[mes]:9,.0f}-{percentil_95[mes]:9,.0f}')

# Tendencia general
print(f'\\nVENTA PROMEDIO PROYECTADA PARA LOS 12 MESES: {np.mean(media_proyectada):,.2f}')
print(f'Variabilidad esperada (desviacion): {np.mean(desv_proyectada):,.2f}')

# Probabilidad de que algunos meses tengan ventas muy bajas
meses_bajos = np.sum(percentil_5 < media_ingresos * 0.5)
print(f'\\nMeses donde hay posibilidad (5%) de ventas muy bajas (<50% del promedio): {meses_bajos}')

# Crecimiento esperado
crecimiento_esperado = ((np.mean(media_proyectada[-3:]) - np.mean(media_proyectada[:3])) / np.mean(media_proyectada[:3])) * 100
print(f'Crecimiento esperado (comparando ultimos 3 vs primeros 3 meses): {crecimiento_esperado:+.2f}%')
```
